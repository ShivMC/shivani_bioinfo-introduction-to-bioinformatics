# Sequence Analysis

This notebook covers fundamental sequence analysis workflows including translation, transcription, and basic statistics.

```python
from Bio.Seq import Seq
from Bio.SeqUtils import GC, molecular_weight
from Bio import SeqIO
```

```python
# Load and analyze a sequence
dna_seq = Seq("ATGCGTACGATCGATCGATCGATCGTAGCTAGCTAGCTAGCTAGC")
print(f"DNA: {dna_seq}")
print(f"RNA: {dna_seq.transcribe()}")
print(f"Protein: {dna_seq.translate()}")
print(f"GC Content: {GC(dna_seq):.2f}%")
```

```python
# Reverse complement
print(f"Reverse Complement: {dna_seq.reverse_complement()}")
```
