# BLAST Demonstration

This notebook demonstrates how to run BLAST searches using Python and the NCBI BLAST API.

```python
# Import required libraries
from Bio.Blast import NCBIWWW, NCBIXML
from Bio import SeqIO
import pandas as pd
```

```python
# Example: Run a BLAST search
sequence = "ATGCGTACGATCGATCGATCGATCGTAGCTAGCTAGCTAGCTAGC"
result_handle = NCBIWWW.qblast("blastn", "nt", sequence)
blast_records = NCBIXML.parse(result_handle)
```

```python
# Parse and display results
for record in blast_records:
    for alignment in record.alignments[:5]:
        for hsp in alignment.hsps[:1]:
            print(f"Sequence: {alignment.title}")
            print(f"Length: {alignment.length}")
            print(f"E-value: {hsp.expect}")
            print(f"Identity: {hsp.identities}/{hsp.align_length}")
            print("-" * 50)
```
