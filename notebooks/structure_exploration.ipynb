# Structure Exploration

This notebook demonstrates how to fetch and analyze protein structures from the PDB.

```python
import Bio.PDB
from Bio.PDB import PDBList, PDBParser
import nglview as nv
```

```python
# Download a structure from PDB
pdbl = PDBList()
pdbl.retrieve_pdb_file("1BNA", pdir="./structures", file_format="pdb")
```

```python
# Parse and analyze structure
parser = PDBParser()
structure = parser.get_structure("1BNA", "./structures/pdb1bna.ent")

for model in structure:
    for chain in model:
        print(f"Chain: {chain.id}, Residues: {len(chain)}")
```
